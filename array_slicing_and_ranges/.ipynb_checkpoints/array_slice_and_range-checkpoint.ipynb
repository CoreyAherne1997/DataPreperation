```python
temperatures = pd.DataFrame({

 'city': ['New York', 'London', 'Paris', 'Berlin', 'Amsterdam'],
 'day1': [23, 25, 27, 26, 24],
 'day2': [22, 21, 25, 26, 23],
 'day3': [26, 25, 24, 27, 23],
 'day4': [23, 21, 22, 26, 27],
 'day5': [27, 26, 27, 24, 28]
})
```

```python
temperatures
```

```python
#how would we deal with this
#we leave the city column alone, turn the day1 and day2 and such into variables of there own
temperatures.melt(id_vars=["city"] , var_name="days" , value_name="temps")
```

```python
stocks=price_df[["date", "symbol" , "close"]]
```

```python
stocks
```

```python
stocks= stocks.set_index("date")
```

```python
stocks
```

```python
stocks = stocks.sort_index()
```

```python
stocks_small = stocks.loc["2016-02-01": "2016-03-01"]
```

```python
stocks_small.reset_index()
```

```python
stocks_small = stocks_small.reset_index()
ssm = stocks_small.pivot(index="symbol", columns="date", values="close").reset_index()
```

```python
ssm
```

```python
ssm.melt(id_vars="symbol",var_name="date",value_name="close")
```
