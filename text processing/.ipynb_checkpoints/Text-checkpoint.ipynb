```python
train = pd.read_csv("train_E6oV3lV.csv")
```

```python
train.shape
```

```python
train.head()
```

```python
train["tweet"].iloc[1]
```

```python
train["word_count"]=train["tweet"].apply(lambda x: len(str(x).split()))
```

```python
train[["tweet","word_count"]].head()
```

```python
train["char_count"]=train["tweet"].str.len()
```

```python
train[["tweet","char_count"]].head()
```

```python
def avg_word(text):
    words=text.split()
    return(sum(len(word) for word in words)/len(words))
```

```python
train["avg_word"]=train["tweet"].apply(lambda x: avg_word(x))
```

```python
train[["tweet","avg_word"]].head()
```

```python
import nltk
nltk.download("stopwords")
```

```python
from nltk.corpus import stopwords
stop = stopwords.words("english")
```

```python
#stop
```

```python
#print(stopwords.fileids())
```

```python
train["stopwords"]=train["tweet"].apply(lambda x: len([x for x in x.split() if x in stop]))
```

```python
train[["tweet","stopwords"]].head()
```

```python
train["hashtags"]=train["tweet"].apply(lambda x: len([x for x in x.split() if x.startswith("#")]))
```

```python
train[["tweet","hashtags"]].head()
```

```python
train["at_sym"]=train["tweet"].apply(lambda x: len([x for x in x.split() if x.startswith("@")]))
```

```python
train[["tweet","at_sym"]].head()
```

```python
train["numeric"]=train["tweet"].apply(lambda x: len([x for x in x.split() if x.isdigit()]))
```

```python
train[["tweet","numeric"]].head()
```

```python
#train["numeric"].unique()
```

```python
train["upper"]=train["tweet"].apply(lambda x: len([x for x in x.split() if x.isupper()]))
```

```python
train[["tweet","upper"]].head()
```

```python
train["upper"].unique()
```

```python
train.head()
```

```python
train["clean_tweet"]=train["tweet"].apply(lambda x: x.lower())
```

```python
train[["tweet","clean_tweet"]].head()
```

```python
train["clean_tweet"]=train["clean_tweet"].str.replace(r"[^\w\s]","", regex=True)
```

```python
train[["tweet","clean_tweet"]].head()
```

```python
train["clean_tweet"]=train["clean_tweet"].apply(lambda x: " ".join(x for x in x.split() if x not in stop))
```

```python
train[["tweet","clean_tweet"]].head()
```

```python
freq = pd.Series(" ".join(train["clean_tweet"]).split()).value_counts()[:10]
```

```python
freq
```

```python
freq=list(freq.index)
```

```python
train["clean_tweet_high_freq"]=train["clean_tweet"].apply(lambda x: " ".join(x for x in x.split() if x not in freq))
```

```python
train[["clean_tweet","clean_tweet_high_freq"]].head()
```
