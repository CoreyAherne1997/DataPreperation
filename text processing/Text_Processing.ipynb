```python
# We want to take these pieces of text and be able to extract things from them.
# Each tweet is considered a single string/single piece of text.
# Hashtag is typically the subject matter, @ is the person talking or someone being
# talked about.
# We only see these in social networks

# @ is typically who and hashtag is typically what

```

```python
#we are gonna extract numerical features from our text
# first we are gonna do a word count, split by the spaces
# put all the words into a list and then get the size of the list

train["word_count"] = train["tweet"].apply(lambda x: len(str(x).split()))
#this will be a new column in our dataframe
```

```python
train[["tweet", "word_count"]].head()
```

```python
#character count

train["char_count"] = train["tweet"].str.len()
```

```python
train[["tweet", "char_count"]].head()
```

```python
#get the average word length
#we need to find the len of every word, add them together and devide by the count

def avg_word(text):
    words=text.split()
    return(sum(len(word) for word in words)/len(words))
```

```python
train["avg_words"] = train["tweet"].apply(lambda x: avg_word(x))
```

```python
train[["tweet", "avg_words"]].head()
```

```python
#index 4 is artificial because the avg words in 8 on the dot, means we need to
#look at it in more depth

```

```python
# next lesson is on grammar and sythex
# the sun is hot and sun hot are the same
# the and is are stop words, joining them together
# they dont have any information gain
# these words cause issue as they pop up alot the machine learning will get biased
# we will remove all the stop words
# first thing we will do is count the stop words
# There are tools, natural language toolkit, that have stop words

import nltk
nltk.download("stopwords")
```

```python
from nltk.corpus import stopwords
stop = stopwords.words("english")
```

```python
#stop
```

```python
train["stopwords"] = train["tweet"].apply(lambda x: len([x for x in x.split() if x in stop]))
```

```python
train[["tweet", "stopwords"]].head()
```

```python
# We all do for social media data, never for an email or letter or noval
# 3 characteristics with social media data, first is the #, @ and upper and lowercase
#we start with getting the #
train["hashtags"] = train["tweet"].apply(lambda x: len([x for x in x.split() if x.startswith("#")]))
```

```python
train[["tweet", "hashtags"]].head()
```

```python
#at_symbol
train["at_symbol"] = train["tweet"].apply(lambda x: len([x for x in x.split() if x.startswith("@")]))
```

```python
train[["tweet", "at_symbol"]].head()
```

```python
#upper cases
train["upper"] = train["tweet"].apply(lambda x: len([x for x in x.split() if x.isupper()]))
```

```python
train[["tweet", "upper"]].head()
```

```python
train["upper"].unique()
```

```python
# spam usually has digits, we can use this 
train["digits"] = train["tweet"].apply(lambda x: len([x for x in x.split() if x.isdigit()]))
```

```python
train[["tweet", "digits"]].head()
```

```python
train["digits"].unique()
```

```python
#4 or 6 digits immediately look at those
train.head()
```

```python
#next we will create a clean dataset
#by doing 3 things, we do to every single word based dataset we get:

#1 ensure everything is in lowercase
train["tweet_clean"] = train["tweet"].apply(lambda x: " ".join(x.lower() for x in x.split()))

#this will get rid of special characters but will keep digits and any spaces. 
#using Regex ^ = special charcaters. w = words, s = string
train["tweet_clean"] =  train["tweet_clean"].str.replace(r"[^\w\s]","", regex=True)
train.head()
train["tweet_clean"] = train["tweet_clean"].str.replace("ð","")
```

```python
#2 getting rid of stopwords, because they introduce noise to the data
from nltk.corpus import stopwords
stop = stopwords.words("english")

train["tweet_clean"] = train["tweet_clean"].apply(lambda x: " ".join(x for x in x.split() if x not in stop)) 

```

```python

train[["tweet", "tweet_clean"]].head()
```

```python
#two other types of words that might not be any use to us
#make a list of the top 5 most common words
top_freq = pd.Series(" ".join(train["tweet_clean"]).split()).value_counts()[:10]
```

```python
top_freq
```

```python
#user is the most freq word, we may want to remove it

#bottom freq
bottom_freq = pd.Series(" ".join(train["tweet_clean"]).split()).value_counts()[-10:]
```

```python
bottom_freq
```

```python
train["tweet_clean"] = train["tweet_clean"].apply(lambda x: " ".join(x for x in x.split() if x not in bottom_freq))
#removed the bottom freq

#predictive text, nearly impossible to make a combination of misspelled words
#take the top 5 tweets and run a model called text blob, see if it can fix the 
#spelling mistakes
```
