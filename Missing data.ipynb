```python
##in all of these different cases where n.a or nan or ? or whatever
#We are gonna use those values by ignoring them to become missing values
#mdl = missing data list
#in other cases it will be other values, this list is for this perticular df
mdl = ["n.a.", "?", "NA", "n/a", "na" , "--"]
# as the data is being read in, the values from our list will be ignored
df = pd.read_csv("employees.csv", na_values = mdl)
#we created a list with the wrong inputs but in by the user
```

```python
df.isnull().sum()
#we can see the number of missing values has increased
```

```python
df.info()
```

```python
#now we have a dataset with aload of missing data
#in 3 ways we can deal with it

#1 drop the missing values:
#take a look at the statistical values before we do that 
df.describe()
```

```python
new_df = df.dropna(axis=0)
# getting rid of the rows that have na
#axis = 0 means rows

new_df.isnull().sum()
#we have no missing values
```

```python
new_df.shape
```

```python
#weve lost observations
new_df.describe()
```

```python
#we can see the values have changed because we dropped rows

#2 flagging:
#We can put in 0 to show there is missing data, in categorical features like gender, we may decide to wrote unknown

new_df = df
new_df.shape
```

```python
new_df["Salary"] = new_df["Salary"].fillna(0)
#we filled the na with 0
```

```python
new_df["Salary"].head()
```

```python
#we must check how that changed the statistical nature of the data 
new_df["Salary"].describe()
```

```python
#fill 0 is the most away from the original values
#we can flag categorical data too
new_df["Gender"]=new_df["Gender"].fillna("Unknown")
new_df["Gender"].head(10)
```

```python
#3 imputation means to put values in somewhere
#in stats of something is normally disprutated, the mean mode and whatever are the same
#never ever use the mean value full stop, the median is a better representation.
#fill with the median, or backfill take value of behind and use that or foward fill use the value after it
new_df = df

```

```python
#forward fill
                                    #here
new_df["Salary"]= new_df["Salary"].ffill()
```

```python
new_df["Salary"].head(10)
```

```python
new_df.describe()
```

```python
#backfill
new_df["Salary"]= new_df["Salary"].bfill()
new_df.describe()
```

```python
#interpolate
new_df["Salary"] = new_df["Salary"].interpolate(method="linear", direction="forward")
new_df.describe()
```
